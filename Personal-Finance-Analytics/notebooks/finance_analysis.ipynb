## Key Insights

1. **Income and Savings**  
   Monthly income demonstrated a strong positive relationship with actual savings, with a correlation of **0.741**. This indicates that higher-income records generally tend to have higher savings in the dataset.

2. **Expenses and Savings**  
   Monthly expenses showed a moderate negative relationship with actual savings, with a correlation of **-0.600**. Records with higher expenses generally tend to have lower savings.

3. **Discretionary Spending and Financial Stress**  
   Average discretionary spending was relatively similar across low, medium, and high financial-stress groups. Therefore, the dataset does not show a strong difference in discretionary spending across stress levels.

4. **Debt-to-Income Ratio and Financial Stress**  
   The average debt-to-income ratio was **0.357** for the high-stress group and **0.351** for the low-stress group. The small difference suggests no strong relationship between debt-to-income ratio and financial stress in this dataset.

5. **Income Type and Savings Goal Achievement**  
   The savings-goal achievement rate was **11.15% for Freelance**, **9.78% for Mixed**, and **8.68% for Salary** income types. Freelance records had the highest observed achievement rate, although this does not establish a causal relationship.

6. **Financial Scenario and Actual Savings**  
   Average actual savings were approximately **₹1,176 under Normal conditions**, **₹1,140 during Inflation**, and **₹1,117 during Recession**. The dataset shows the highest average savings under Normal conditions and the lowest under Recession conditions.

7. **Emergency Fund and Financial Stress**  
   Average emergency funds were approximately **₹1,018 for Low stress**, **₹994 for Medium stress**, and **₹989 for High stress**. The relatively small differences indicate no strong variation in emergency funds across stress levels.

8. **Subscription Services and Savings**  
   The number of subscription services did not show a consistent increasing or decreasing pattern with actual savings. This suggests that subscription count alone does not clearly explain savings differences in the dataset.

9. **Cash Flow Status and Savings Goal Achievement**  
   The savings-goal achievement rate was **8.72% for Negative cash flow**, **11.11% for Neutral cash flow**, and **8.75% for Positive cash flow**. No clear relationship between cash-flow status and goal achievement was observed.

10. **Fraud Rate by Expense Category**  
    The highest observed fraud rates were recorded in **Groceries (3.25%)** and **Rent (3.01%)**, while **Dining Out (1.31%)** had the lowest observed rate. These figures represent patterns within this dataset and do not imply that any category is inherently more susceptible to fraud.

11. **Category-wise Average Expenses**  
    Average monthly expenses were highest for **Groceries (₹3,076)** and lowest for **Rent (₹2,929)**. The difference of approximately **₹147** indicates that average expenses were relatively similar across the analyzed categories.
